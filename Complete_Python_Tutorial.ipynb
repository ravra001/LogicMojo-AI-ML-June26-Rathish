## 2. Syntax & Statements

Python's syntax relies on a few core rules:

- **Indentation defines blocks.** Code inside `if`, `for`, `def`, etc. must be indented
  consistently (4 spaces is the convention). Mixing tabs and spaces causes errors.
- **A statement** is normally one line of code. You can put multiple statements on
  one line using a semicolon `;`, though this is discouraged for readability.
- **Line continuation** — a long statement can be split across lines using a
  backslash `\`, or implicitly inside brackets `()`, `[]`, `{}`.

````python
# Indentation defines a block
if 5 > 2:
    print("Five is greater than two!")   # indented = inside the if-block

# Multiple statements on one line (semicolon) — generally avoid in real code
a = 1; b = 2; print(a + b)

# Line continuation with backslash
total = 1 + 2 + 3 + \
        4 + 5
print(total)

# Implicit continuation inside brackets (preferred style)
numbers = [
    1, 2, 3,
    4, 5, 6
]
print(numbers)
````

## 5. Variables

A variable is created the moment you first assign a value to it — there's no
separate "declare" step.

**Naming rules:**
- Must start with a letter or underscore (not a digit).
- Can contain letters, digits, and underscores.
- Case-sensitive (`age` and `Age` are different variables).
- Cannot be a reserved keyword (e.g. `class`, `for`, `if`).

**Other variable features:**
- **Multiple assignment** — assign several variables in one line.
- **Output variables** — combine with `print()` using commas, `+` (strings only), or f-strings.
- **Global variables** — created outside any function; accessible everywhere unless
  shadowed by a local variable of the same name inside a function.

````python
# Basic assignment
x = 5
name = "Alice"
print(x, name)

# Variable naming
_valid_name = 1
camelCase = 2
snake_case = 3
print(_valid_name, camelCase, snake_case)

# Assigning multiple values at once
a, b, c = 1, 2, 3
print(a, b, c)

# Assign the same value to multiple variables
x = y = z = "Same"
print(x, y, z)

# Output variables
age = 30
print("Age is", age)            # comma -> auto adds a space
print("Age is " + str(age))     # + needs both sides to be strings
print(f"Age is {age}")          # f-string (recommended, modern style)

# Global variables
message = "I am global"

def show_message():
    print(message)   # can read the global variable inside a function

show_message()

def change_message():
    global message    # 'global' keyword lets us MODIFY the global variable
    message = "Changed inside function"

change_message()
print(message)
````

````python
# str -> int / float
print(int("10"))
print(float("3.5"))

# int -> float / str
print(float(10))
print(str(10))

# float -> int (drops the decimal part, doesn't round)
print(int(7.9))

# Casting can raise an error if the value isn't a valid number
try:
    int("hello")
except ValueError as e:
    print("Error:", e)
````

## 9. Strings

Strings are sequences of characters enclosed in single or double quotes. Strings
support indexing, slicing, and a large set of built-in methods.

- **Slicing** — `string[start:end:step]` extracts a substring.
- **Modifying** — strings are *immutable*; methods like `.upper()`, `.lower()`,
  `.strip()`, `.replace()` return a *new* string rather than changing the original.
- **Concatenation** — joining strings with `+`.
- **Formatting** — f-strings (`f"..."`) or `.format()` for inserting values.
- **Escape characters** — backslash sequences like `\n`, `\t`, `\\`, `\"` for
  special characters inside a string.

````python
s = "Hello, World!"

# Basic indexing (0-based)
print(s[0])      # 'H'
print(s[-1])     # '!' (negative index counts from the end)

# Slicing string[start:end:step]
print(s[0:5])    # 'Hello'
print(s[7:])     # 'World!'
print(s[:5])     # 'Hello'
print(s[::-1])   # reversed string

# Modifying (returns new strings; original is unchanged)
print(s.upper())
print(s.lower())
print("  padded  ".strip())
print(s.replace("World", "Python"))

# Concatenation
greeting = "Hello" + ", " + "World!"
print(greeting)

# Formatting
name, age = "Alice", 30
print(f"{name} is {age} years old")          # f-string
print("{} is {} years old".format(name, age)) # .format()

# Escape characters
print("Line1\nLine2")     # newline
print("Tab\tSeparated")   # tab
print("She said \"Hi\"") # escaped quote
print("Backslash: \\")   # literal backslash

# A few common string methods
text = "python is fun"
print(text.split())          # -> ['python', 'is', 'fun']
print(text.title())          # -> 'Python Is Fun'
print(len(text))             # length of string
print("fun" in text)         # membership check
print(text.find("is"))       # index of substring
````

## 10. Booleans

Booleans represent one of two values: `True` or `False`. They're the result of
comparison operations, and almost any value can be evaluated for "truthiness"
using `bool()`.

**Falsy values** in Python: `False`, `0`, `0.0`, `""` (empty string), `[]`, `{}`,
`()`, `set()`, and `None`. Everything else is truthy.

````python
print(10 > 9)     # True
print(10 == 9)    # False
print(10 < 9)     # False

x = bool(1)        # True
y = bool(0)        # False
print(x, y)

# Truthy / falsy checks
print(bool("hello"))   # True (non-empty string)
print(bool(""))        # False (empty string)
print(bool([1, 2]))     # True (non-empty list)
print(bool([]))        # False (empty list)

if []:
    print("This won't print")
else:
    print("Empty list is falsy")
````

## 11. Operators

Python groups operators into several categories:

- **Arithmetic**: `+ - * / % ** //`
- **Assignment**: `= += -= *= /= %= **= //=` etc.
- **Comparison**: `== != > < >= <=`
- **Logical**: `and or not`
- **Identity**: `is`, `is not` (compares object identity, not value)
- **Membership**: `in`, `not in`
- **Bitwise**: `& | ^ ~ << >>`
- **Ternary (conditional expression)**: `value_if_true if condition else value_if_false`

A quick note on **operator precedence**: parentheses `()` first, then `**`, then
`* / // %`, then `+ -`, then comparisons, then `not`, `and`, `or`.

````python
# Arithmetic
print(10 + 3, 10 - 3, 10 * 3, 10 / 3, 10 % 3, 10 ** 2, 10 // 3)

# Assignment operators
x = 5
x += 3   # same as x = x + 3
print(x)

# Comparison
print(5 == 5, 5 != 4, 5 > 3, 5 <= 5)

# Logical
a, b = True, False
print(a and b, a or b, not a)

# Identity vs equality
list1 = [1, 2, 3]
list2 = [1, 2, 3]
print(list1 == list2)   # True -> same VALUES
print(list1 is list2)   # False -> different OBJECTS in memory

# Membership
fruits = ["apple", "banana"]
print("apple" in fruits)
print("grape" not in fruits)

# Bitwise
print(5 & 3, 5 | 3, 5 ^ 3, ~5, 5 << 1, 5 >> 1)

# Ternary operator
age = 20
status = "adult" if age >= 18 else "minor"
print(status)

# Operator precedence example
print(2 + 3 * 4)     # 14, not 20 (multiplication first)
print((2 + 3) * 4)   # 20 (parentheses override precedence)
````

## 12. Lists

A list is an ordered, mutable, indexable collection that allows duplicate values.

Covers: **access** (indexing/slicing), **change**, **add** (`append`, `insert`,
`extend`), **remove** (`remove`, `pop`, `del`, `clear`), **loop** (`for`, `while`,
list comprehension), **sort** (`sort`, `sorted`, `reverse`), **copy**
(`.copy()`, `list()`), and **join** (`+`, `extend`).

````python
fruits = ["apple", "banana", "cherry"]

# Access
print(fruits[0])      # 'apple'
print(fruits[-1])     # 'cherry'
print(fruits[0:2])    # slicing -> ['apple', 'banana']

# Change
fruits[1] = "blueberry"
print(fruits)

# Add items
fruits.append("date")          # add to the end
fruits.insert(1, "kiwi")       # insert at a specific index
print(fruits)

# Remove items
fruits.remove("kiwi")          # removes by value
popped = fruits.pop()          # removes & returns the last item
print("Popped:", popped, "->", fruits)
del fruits[0]                  # removes by index
print(fruits)

# Loop through a list
for fruit in fruits:
    print("Fruit:", fruit)

# List comprehension (compact way to build a list)
squares = [x**2 for x in range(6)]
print(squares)
evens = [x for x in range(10) if x % 2 == 0]
print(evens)

# Sort
numbers = [5, 2, 9, 1]
numbers.sort()                    # sorts in place
print(numbers)
print(sorted(numbers, reverse=True))  # returns a NEW sorted list

# Copy (use .copy() or list(), NOT new_list = old_list, which just creates a reference)
original = [1, 2, 3]
copy1 = original.copy()
copy2 = list(original)
copy1.append(4)
print(original, copy1, copy2)

# Join lists
list_a = [1, 2]
list_b = [3, 4]
print(list_a + list_b)         # concatenation
list_a.extend(list_b)          # extend in place
print(list_a)
````

## 13. Tuples

A tuple is an ordered, **immutable** collection that allows duplicates. Because
tuples can't be changed after creation, "updating" one means converting it to a
list, modifying it, then converting back.

Covers: **access**, **update (workaround)**, **unpacking**, **loop**, **join**,
and **methods** (`count`, `index`).

````python
colors = ("red", "green", "blue")

# Access
print(colors[0])
print(colors[-1])
print(colors[0:2])

# "Update" a tuple (tuples are immutable, so convert to a list first)
temp = list(colors)
temp[1] = "yellow"
colors = tuple(temp)
print(colors)

# Unpacking
a, b, c = colors
print(a, b, c)

first, *rest = colors          # 'rest' collects the remaining items
print(first, rest)

# Loop through a tuple
for color in colors:
    print("Color:", color)

# Join tuples
tuple_a = (1, 2)
tuple_b = (3, 4)
print(tuple_a + tuple_b)

# Tuple methods
nums = (1, 2, 2, 3, 2)
print(nums.count(2))   # how many times 2 appears
print(nums.index(3))   # index of first occurrence of 3
````

## 14. Sets

A set is an **unordered**, **unindexed** collection that does not allow duplicates.
Because sets are unordered, you can't access items by index — you loop through
them or check membership with `in`.

Covers: **add**, **remove** (`remove`, `discard`, `pop`), **loop**, **join**
(`union`, `update`), **frozenset** (immutable set), and common **methods**.

````python
fruits = {"apple", "banana", "cherry"}
print(fruits)

# Access (no indexing -- loop or use 'in')
print("apple" in fruits)
for fruit in fruits:
    print("Fruit:", fruit)

# Add items
fruits.add("date")
fruits.update(["elderberry", "fig"])   # add multiple items
print(fruits)

# Remove items
fruits.remove("date")     # raises an error if not found
fruits.discard("xyz")     # does NOT raise an error if not found
popped = fruits.pop()     # removes a random item (sets are unordered)
print(fruits)

# Join sets
set_a = {1, 2, 3}
set_b = {3, 4, 5}
print(set_a.union(set_b))          # all unique items from both
print(set_a.intersection(set_b))   # items in both
print(set_a.difference(set_b))     # items only in set_a
set_a.update(set_b)                # merges set_b into set_a
print(set_a)

# Frozenset (immutable set)
fs = frozenset([1, 2, 3])
print(fs)
try:
    fs.add(4)
except AttributeError as e:
    print("Error:", e)
````

## 15. Dictionaries

A dictionary stores data as **key-value pairs**. Keys must be unique and
immutable (strings, numbers, tuples); values can be anything.

Covers: **access** (`[]`, `.get()`), **change**, **add**, **remove** (`pop`,
`popitem`, `del`, `clear`), **loop** (`.keys()`, `.values()`, `.items()`),
**copy**, **nested dictionaries**, and **methods**.

````python
person = {"name": "Alice", "age": 30, "city": "Chennai"}

# Access
print(person["name"])
print(person.get("age"))
print(person.get("country", "Not specified"))   # default if key missing

# Change
person["age"] = 31
print(person)

# Add a new key-value pair
person["email"] = "alice@example.com"
print(person)

# Remove items
person.pop("email")
print(person)
removed_key, removed_val = list(person.items())[-1]
del person["city"]
print(person)

# Loop through a dictionary
for key in person.keys():
    print("Key:", key)
for value in person.values():
    print("Value:", value)
for key, value in person.items():
    print(f"{key} -> {value}")

# Copy
person_copy = person.copy()
person_copy["age"] = 99
print(person, person_copy)   # original is unaffected

# Nested dictionaries
people = {
    "person1": {"name": "Alice", "age": 30},
    "person2": {"name": "Bob", "age": 25}
}
print(people["person1"]["name"])

# Dictionary methods
print(len(person))
print("name" in person)
person.update({"phone": "555-1234"})
print(person)
````

## 16. If...Else (Conditionals)

Conditional statements run different code depending on whether a condition is
`True` or `False`.

Covers: **if**, **elif**, **else**, **shorthand if** (ternary, one-liner),
**logical operators** in conditions, **nested if**, and the **pass** statement
(used as a placeholder when a block is syntactically required but you don't
want it to do anything yet).

````python
age = 20

# if / elif / else
if age < 13:
    print("Child")
elif age < 20:
    print("Teenager")
else:
    print("Adult")

# Shorthand if (single statement)
if age >= 18: print("Can vote")

# Ternary / conditional expression
status = "adult" if age >= 18 else "minor"
print(status)

# Logical operators in conditions
has_id = True
if age >= 18 and has_id:
    print("Entry allowed")

# Nested if
num = 15
if num > 0:
    if num % 2 == 0:
        print("Positive even number")
    else:
        print("Positive odd number")

# pass statement (placeholder, does nothing)
if age > 100:
    pass   # TODO: handle this case later
else:
    print("Normal age range")
````

## 17. Match Statement

`match` (added in Python 3.10) is a structural pattern-matching statement,
similar to `switch` in other languages. It compares a value against several
`case` patterns. The wildcard `_` matches anything (like a default case), and
`|` lets you match multiple values in one case.

````python
def describe_day(day):
    match day:
        case "Saturday" | "Sunday":
            return "Weekend"
        case "Monday":
            return "Start of the work week"
        case _:
            return "A regular weekday"

print(describe_day("Sunday"))
print(describe_day("Monday"))
print(describe_day("Wednesday"))

# match can also work with values and simple structures
def http_status(code):
    match code:
        case 200:
            return "OK"
        case 404:
            return "Not Found"
        case 500:
            return "Server Error"
        case _:
            return "Unknown status"

print(http_status(404))
````

## 18. While Loops

A `while` loop repeats as long as a condition is `True`. Use `break` to exit
early, `continue` to skip to the next iteration, and an optional `else` clause
that runs once when the loop condition becomes `False` (but not if you `break`).

````python
i = 1
while i <= 5:
    print(i)
    i += 1

# break -- stop the loop entirely
i = 1
while i <= 10:
    if i == 4:
        break
    print("break demo:", i)
    i += 1

# continue -- skip the rest of this iteration
i = 0
while i < 5:
    i += 1
    if i == 3:
        continue
    print("continue demo:", i)

# else clause -- runs when the condition becomes False (not on break)
i = 1
while i <= 3:
    print("loop:", i)
    i += 1
else:
    print("Loop finished normally")
````

## 19. For Loops

A `for` loop iterates over a sequence (list, tuple, string, dict, set, or
range). Like `while`, it supports `break`, `continue`, and an `else` clause.
Loops can also be **nested**.

````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)

# Looping with range()
for i in range(5):
    print("i =", i)

# break and continue
for i in range(10):
    if i == 5:
        break
    if i % 2 == 0:
        continue
    print("odd number:", i)

# else clause (runs if the loop completes without a break)
for i in range(3):
    print("checking", i)
else:
    print("Loop completed fully")

# Nested loops
for i in range(1, 3):
    for j in range(1, 3):
        print(f"i={i}, j={j}")
````

````python
# *args -- variable number of positional arguments (collected into a tuple)
def add_all(*args):
    return sum(args)

print(add_all(1, 2, 3, 4))
````

### Variable Keyword Arguments (`**kwargs`)

````python
# **kwargs -- variable number of keyword arguments (collected into a dict)
def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_info(name="Bob", age=25)
````

### Function Scope (`global`, `local`, `nonlocal`)

````python
# Scope
x = "global value"

def show_scope():
    x = "local value"   # this is a NEW local variable, separate from the global one
    print("Inside function:", x)

show_scope()
print("Outside function:", x)

def outer():
    y = "outer value"
    def inner():
        nonlocal y           # lets inner() modify outer()'s variable
        y = "modified by inner"
    inner()
    print(y)

outer()
````

### Decorators

````python
# Decorators
def my_decorator(func):
    def wrapper():
        print("Something before the function runs")
        func()
        print("Something after the function runs")
    return wrapper

@my_decorator
def say_hello():
    print("Hello!")

say_hello()
````

### Lambda Functions

````python
# Lambda functions
square = lambda x: x ** 2
print(square(5))

add = lambda a, b: a + b
print(add(3, 4))
````

### Recursion

````python
# Recursion
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
````

### Generators

````python
# Generators
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

for number in count_up_to(5):
    print("generated:", number)
````

````python
# Basic function with a default argument
def greet(name="friend"):
    return f"Hello, {name}!"

print(greet())
print(greet("Alice"))

# Keyword arguments
def describe_pet(name, animal="dog"):
    print(f"{name} is a {animal}")

describe_pet(animal="cat", name="Whiskers")   # order doesn't matter with keywords

# *args -- variable number of positional arguments (collected into a tuple)
def add_all(*args):
    return sum(args)

print(add_all(1, 2, 3, 4))

# **kwargs -- variable number of keyword arguments (collected into a dict)
def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_info(name="Bob", age=25)

# Scope
x = "global value"

def show_scope():
    x = "local value"   # this is a NEW local variable, separate from the global one
    print("Inside function:", x)

show_scope()
print("Outside function:", x)

def outer():
    y = "outer value"
    def inner():
        nonlocal y           # lets inner() modify outer()'s variable
        y = "modified by inner"
    inner()
    print(y)

outer()

# Decorators
def my_decorator(func):
    def wrapper():
        print("Something before the function runs")
        func()
        print("Something after the function runs")
    return wrapper

@my_decorator
def say_hello():
    print("Hello!")

say_hello()

# Lambda functions
square = lambda x: x ** 2
print(square(5))

add = lambda a, b: a + b
print(add(3, 4))

# Recursion
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))

# Generators
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

for number in count_up_to(5):
    print("generated:", number)
````

## 21. Range

`range(start, stop, step)` generates a sequence of numbers, commonly used in
`for` loops. `stop` is required; `start` defaults to 0 and `step` defaults to 1.
The `stop` value is **never included**.

````python
print(list(range(5)))           # [0, 1, 2, 3, 4]
print(list(range(2, 8)))        # [2, 3, 4, 5, 6, 7]
print(list(range(0, 10, 2)))    # [0, 2, 4, 6, 8]
print(list(range(10, 0, -1)))   # counting down

for i in range(3):
    print("range loop:", i)
````

## 23. Iterators

An **iterable** is any object you can loop over (lists, tuples, strings, etc.).
An **iterator** is the object that actually produces the next value, created
from an iterable using `iter()` and advanced with `next()`. You can build a
custom iterator by implementing `__iter__()` and `__next__()` on a class.
When there are no more items, `next()` raises `StopIteration`.

````python
# Built-in iterables -> iterators
my_list = [1, 2, 3]
my_iter = iter(my_list)
print(next(my_iter))
print(next(my_iter))
print(next(my_iter))

# Calling next() again raises StopIteration
try:
    print(next(my_iter))
except StopIteration:
    print("No more items!")

# A custom iterator class
class Counter:
    def __init__(self, limit):
        self.limit = limit
        self.current = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < self.limit:
            self.current += 1
            return self.current
        raise StopIteration

for number in Counter(4):
    print("custom iterator:", number)
````

## 28. RegEx (Regular Expressions)

The `re` module lets you search, match, and manipulate text using patterns.

Common functions:
- `re.search()` — finds the first match anywhere in the string.
- `re.findall()` — returns a list of all matches.
- `re.split()` — splits a string at each match.
- `re.sub()` — replaces matches with another string.

Common special sequences: `\d` (digit), `\w` (word character), `\s` (whitespace),
`^` (start of string), `$` (end of string), `*` (0 or more), `+` (1 or more).

````python
import re

text = "My phone numbers are 123-456-7890 and 987-654-3210"

# search -- find the first match
match = re.search(r"\d{3}-\d{3}-\d{4}", text)
print(match.group())

# findall -- find every match
all_numbers = re.findall(r"\d{3}-\d{3}-\d{4}", text)
print(all_numbers)

# split -- split string on a pattern
words = re.split(r"\s+", "Python   is    fun")
print(words)

# sub -- replace matches
masked = re.sub(r"\d{3}-\d{3}-\d{4}", "[REDACTED]", text)
print(masked)

# Common patterns
print(re.findall(r"\w+", "Hello, World! 123"))   # word characters
print(re.match(r"^Hello", "Hello, World!"))        # anchored at start
````

## 30. Try...Except

Exception handling lets your program respond to errors gracefully instead of
crashing.

- `try` — code that might raise an error.
- `except` — runs if an error occurs (can target specific exception types).
- `else` — runs if NO error occurred.
- `finally` — always runs, error or not (great for cleanup).
- `raise` — manually trigger an exception.

````python
# Basic try/except
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print("Error caught:", e)

# Catching multiple / specific exception types
try:
    value = int("not a number")
except ValueError:
    print("That wasn't a valid number")
except TypeError:
    print("Wrong type")

# else and finally
try:
    x = 10 / 2
except ZeroDivisionError:
    print("Division error")
else:
    print("No errors occurred, result:", x)
finally:
    print("This always runs (cleanup, closing resources, etc.)")

# Manually raising an exception
def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    return age

try:
    check_age(-5)
except ValueError as e:
    print("Caught custom error:", e)
````

## 31. String Formatting

Beyond basic f-strings, Python supports format specifiers for controlling

````python
price = 49.99
quantity = 3
pi = 3.14159265

# Basic f-string
print(f"Price: {price}")

# Decimal places
print(f"Pi rounded: {pi:.2f}")        # 2 decimal places

# Padding / width and alignment
print(f"[{quantity:5}]")              # right-align in width 5
print(f"[{quantity:<5}]")             # left-align
print(f"[{quantity:^5}]")             # center-align

# Padding with zeros
print(f"{7:03}")                      # -> 007

# Thousands separator
print(f"{1234567:,}")                 # -> 1,234,567

# Percentage
ratio = 0.856
print(f"{ratio:.1%}")                 # -> 85.6%

# Combining with .format()
print("Total: {:.2f}".format(price * quantity))
````

## 1. Object-Oriented Programming (OOP) — Introduction

OOP organizes code around **objects** that bundle together data (attributes)
and behavior (methods). The four pillars of OOP are:

1. **Encapsulation** — bundling data and methods together, restricting direct
   access to some of an object's internals.
2. **Inheritance** — a class can inherit attributes/methods from another class.
3. **Polymorphism** — different classes can implement the same method name in
   their own way (the same call behaves differently depending on the object).
4. **Abstraction / Class & Object basics** — a **class** is a blueprint; an
   **object** is a specific instance created from that blueprint.

We'll build each of these up step by step below.

## 2. Classes and Objects

A **class** is defined with the `class` keyword and acts as a template. An
**object** is an instance of that class, created by calling the class like a
function.

````python
class Car:
    pass   # an empty class body (placeholder)

# Creating objects (instances) from the class
car1 = Car()
car2 = Car()

# Adding attributes directly to an instance
car1.brand = "Toyota"
car2.brand = "Honda"

print(car1.brand)
print(car2.brand)
print(type(car1))
````

## 3. The `__init__` Method

`__init__` is a special method (a "constructor") that runs automatically when
a new object is created. It's used to set up the object's initial attributes.

````python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

car1 = Car("Toyota", "Corolla")
print(car1.brand, car1.model)

car2 = Car("Honda", "Civic")
print(car2.brand, car2.model)
````

## 4. The `self` Parameter

`self` refers to the *current instance* of the class. It's the first
parameter of every instance method (though you never pass it explicitly —
Python does that automatically), and it's how methods read/write that
specific object's attributes. The name `self` is just a convention; you could
technically call it anything, but virtually all Python code uses `self`.

````python
class Car:
    def __init__(self, brand):
        self.brand = brand   # self.brand belongs to THIS object

    def show_brand(self):
        print(f"This car's brand is {self.brand}")

car1 = Car("Toyota")
car2 = Car("Honda")

car1.show_brand()   # uses car1's own 'self'
car2.show_brand()   # uses car2's own 'self'

# Using a different name instead of 'self' (works, but not conventional)
class Car2:
    def __init__(this, brand):
        this.brand = brand
    def show_brand(this):
        print(this.brand)

Car2("Ford").show_brand()
````

## 5. Class Properties

Properties (attributes) can be **instance attributes** (unique to each
object, usually set in `__init__`) or **class attributes** (shared by all
instances of the class, defined directly inside the class body). You can also
modify or delete an object's properties after creation.

````python
class Car:
    wheels = 4   # class attribute -- shared by ALL Car instances

    def __init__(self, brand, color):
        self.brand = brand   # instance attribute -- unique per object
        self.color = color

car1 = Car("Toyota", "red")
car2 = Car("Honda", "blue")

print(car1.wheels, car2.wheels)   # both share the same class attribute
print(car1.color, car2.color)      # each has its own instance attribute

# Modify a property
car1.color = "black"
print(car1.color)

# Delete a property
del car1.color
try:
    print(car1.color)
except AttributeError as e:
    print("Error:", e)

# Changing the class attribute affects all instances that haven't overridden it
Car.wheels = 6
print(car2.wheels)
````

## 6. Class Methods

Beyond regular instance methods (which take `self`), Python also supports:
- **`@classmethod`** — takes `cls` (the class itself) instead of `self`,
  useful for factory-style construction or operating on class-level data.
- **`@staticmethod`** — takes neither `self` nor `cls`; it's just a regular
  function namespaced inside the class for organizational purposes.

````python
class Car:
    total_cars = 0

    def __init__(self, brand):
        self.brand = brand
        Car.total_cars += 1

    def describe(self):              # regular instance method
        return f"This is a {self.brand}"

    @classmethod
    def from_string(cls, car_string):  # alternate constructor
        brand = car_string.split("-")[0]
        return cls(brand)

    @staticmethod
    def general_info():               # doesn't need self or cls
        return "Cars are a means of transportation"

car1 = Car("Toyota")
car2 = Car.from_string("Honda-Civic")

print(car1.describe())
print(car2.describe())
print(Car.total_cars)
print(Car.general_info())
````

## 7. Inheritance

A **child class** can inherit attributes and methods from a **parent class**,
and can override or extend them. `super()` lets a child call the parent
class's version of a method (commonly `__init__`).

````python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

    def describe(self):
        return f"This vehicle is a {self.brand}"

class Car(Vehicle):           # Car inherits from Vehicle
    def __init__(self, brand, model):
        super().__init__(brand)   # call the parent's __init__
        self.model = model

    def describe(self):            # override the parent's method
        base = super().describe()  # we can still call the parent version
        return f"{base}, model {self.model}"

car1 = Car("Toyota", "Corolla")
print(car1.describe())
print(car1.brand)   # inherited attribute
````

## 8. Polymorphism

Polymorphism means different classes can define a method with the **same
name**, and calling that method behaves differently depending on the actual
object's class — even when looping through a mixed collection of objects.

````python
class Car:
    def move(self):
        return "drives on the road"

class Boat:
    def move(self):
        return "sails on water"

class Plane:
    def move(self):
        return "flies in the sky"

vehicles = [Car(), Boat(), Plane()]

for v in vehicles:
    print(f"{type(v).__name__} {v.move()}")   # same method name, different behavior
````

## 9. Encapsulation

Encapsulation restricts direct access to some of an object's components.
Python uses **naming conventions** rather than strict access control:
- `_single_underscore` — "protected" by convention (internal use, but still accessible).
- `__double_underscore` — "private" by convention (name-mangled, harder to access directly).

The `@property` decorator lets you define **getters and setters** that look
like plain attribute access from the outside, while letting you add
validation logic.

````python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   # "private" attribute (name-mangled)

    @property
    def balance(self):              # getter
        return self.__balance

    @balance.setter
    def balance(self, amount):      # setter (with validation)
        if amount < 0:
            raise ValueError("Balance cannot be negative")
        self.__balance = amount

account = BankAccount(1000)
print(account.balance)        # looks like attribute access, but calls the getter

account.balance = 1500         # calls the setter
print(account.balance)

try:
    account.balance = -50      # setter validation kicks in
except ValueError as e:
    print("Error:", e)

# Directly accessing the "private" attribute is discouraged and name-mangled
print(account._BankAccount__balance)   # technically still reachable, but not the intended API
````

## 10. Inner Classes

A class can be defined **inside another class**. This is useful when the
inner class is only meaningful in the context of the outer class (e.g. an
`Engine` that only makes sense as part of a `Car`).

````python
class Car:
    def __init__(self, brand):
        self.brand = brand
        self.engine = self.Engine()   # create an instance of the inner class

    class Engine:
        def __init__(self):
            self.horsepower = 150

        def start(self):
            return "Engine starting..."

car1 = Car("Toyota")
print(car1.brand)
print(car1.engine.horsepower)
print(car1.engine.start())

# The inner class can also be accessed via the outer class directly
generic_engine = Car.Engine()
print(generic_engine.horsepower)
````

## 1. File Handling — Introduction

Python's built-in `open()` function is the key tool for working with files.
It takes a filename and a **mode**:

| Mode | Meaning |
|---|---|
| `"r"` | Read (default) — file must exist |
| `"a"` | Append — creates the file if it doesn't exist |
| `"w"` | Write — overwrites the file if it exists, creates it if not |
| `"x"` | Create — fails if the file already exists |
| `"t"` | Text mode (default) |
| `"b"` | Binary mode |

Modes can be combined, e.g. `"rb"` (read binary), `"wt"` (write text).

**Best practice:** always use a `with` statement when working with files — it
automatically closes the file for you, even if an error occurs.

````python
# Demonstrating the open() signature (not run standalone -- see Read/Write sections below)
help(open)
````

## 2. Read Files

Once a file is open in read mode, you can read its contents using:
- `.read()` — entire file as one string (optionally `.read(n)` for the first n characters).
- `.readline()` — one line at a time.
- `.readlines()` — all lines, returned as a list of strings.
- Looping directly over the file object — reads line by line, memory-efficient for large files.

````python
# First, create a sample file to read from (covered in detail in the next section)
with open("sample.txt", "w") as f:
    f.write("Line 1: Hello\n")
    f.write("Line 2: This is a sample file\n")
    f.write("Line 3: For demonstrating file reading\n")

# Read the entire file at once
with open("sample.txt", "r") as f:
    content = f.read()
print(content)

# Read only the first 10 characters
with open("sample.txt", "r") as f:
    print(f.read(10))

# Read one line at a time
with open("sample.txt", "r") as f:
    print(f.readline().strip())
    print(f.readline().strip())

# Read all lines into a list
with open("sample.txt", "r") as f:
    lines = f.readlines()
print(lines)

# Loop through the file line by line (memory-efficient for big files)
with open("sample.txt", "r") as f:
    for line in f:
        print("Looping:", line.strip())
````

## 3. Write / Create Files

- `"w"` mode — creates the file if it doesn't exist, **overwrites** it if it does.
- `"a"` mode — creates the file if it doesn't exist, **appends** to the end if it does.
- `"x"` mode — creates a new file, but **raises an error** if the file already exists
  (a safe way to avoid accidentally overwriting something).

````python
# "w" -- write (overwrites existing content)
with open("notes.txt", "w") as f:
    f.write("This is the first line.\n")
    f.write("This is the second line.\n")

with open("notes.txt", "r") as f:
    print("After 'w' mode:")
    print(f.read())

# "a" -- append (adds to the end without erasing existing content)
with open("notes.txt", "a") as f:
    f.write("This line was appended.\n")

with open("notes.txt", "r") as f:
    print("After 'a' mode:")
    print(f.read())

# "x" -- create new file, fails if it already exists
import os
if os.path.exists("new_file.txt"):
    os.remove("new_file.txt")   # clean up from any previous run

with open("new_file.txt", "x") as f:
    f.write("Brand new file created safely.")
print("Created new_file.txt successfully")

# Trying 'x' again on the same file raises an error
try:
    with open("new_file.txt", "x") as f:
        f.write("This won't work")
except FileExistsError as e:
    print("Error:", e)
````

## 4. Delete Files

File and folder deletion isn't part of `open()` — it's handled by the `os`
module:
- `os.remove(path)` — deletes a file.
- `os.path.exists(path)` — checks whether a file/folder exists (good practice
  before deleting, to avoid an error).
- `os.rmdir(path)` — removes an **empty** folder.

````python
import os

# Check before deleting (best practice)
if os.path.exists("new_file.txt"):
    os.remove("new_file.txt")
    print("new_file.txt deleted")
else:
    print("The file does not exist")

# Attempting to delete a file that doesn't exist raises an error
try:
    os.remove("does_not_exist.txt")
except FileNotFoundError as e:
    print("Error:", e)

# Removing an empty folder
os.makedirs("temp_folder", exist_ok=True)
print("temp_folder exists:", os.path.exists("temp_folder"))
os.rmdir("temp_folder")
print("temp_folder exists after removal:", os.path.exists("temp_folder"))

# Clean up the other sample files created in this section
for fname in ["sample.txt", "notes.txt"]:
    if os.path.exists(fname):
        os.remove(fname)
print("Cleanup complete")
````
